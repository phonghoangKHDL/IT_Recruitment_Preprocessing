# Introduction
This notebook documents how the raw dataset Data_Raw.csv was created. In Data Science, recording data provenance is essential to ensure reproducibility and to clearly define the starting point of the preprocessing pipeline.
This notebook covers two main tasks. First, it describes the semi-manual process used to collect 250 IT job postings. Second, it verifies that the raw dataset has not been modified in any way.

## 1. Data Collection
The team did not use automated scraping tools or APIs. Instead, 250 job descriptions were manually extracted from two major IT recruitment platforms in Vietnam, ITviec and TopDev.
Each record contains six fields. Job_Title is the position name. Company is the employer name. Location is the work location. Salary_Raw is the salary information copied exactly as shown on the platform. Link is the original URL of the job posting. JD and Requirements is the full job description and requirements text.
## 2. Immutable Raw Principle
No cleaning or modification was applied during collection. All original formatting, line breaks, and inconsistencies were preserved as-is. This process constitutes data acquisition, not data manipulation, as no transformation or interpretation was performed on the collected content. The resulting file represents the original source of truth and fully satisfies the No Intervention rule.
The Link field serves as an audit trail, allowing the team to detect duplicate records and trace each entry back to its original source.

```python
%pip install pandas
import pandas as pd

# Configure pandas display options for better readability
pd.set_option('display.max_colwidth', None)
pd.set_option('display.max_rows', 10)

print('Data Acquisition environment initialized successfully.')
```

```python
# Load the Raw dataset
raw_data_path = '../Data/01_raw/Data_Raw.csv'

try:
    df_raw = pd.read_csv(raw_data_path)
    print(f'Dataset loaded successfully.')
    print(f'Dataset dimensions: {df_raw.shape[0]} rows, {df_raw.shape[1]} columns.')
    display(df_raw.head(2))
except FileNotFoundError:
    print(f'ERROR: File not found at {raw_data_path}.')
```

```python
# Validate uniqueness via the Link field and Job Title + Company combination

duplicate_links = df_raw.duplicated(subset=['Link'], keep=False)
num_duplicates = duplicate_links.sum()

duplicate_title_company = df_raw.duplicated(subset=['Job_Title', 'Company'], keep=False)
num_duplicates_tc = duplicate_title_company.sum()

if num_duplicates == 0:
    print('No duplicate Link references detected.')
else:
    print(f'Warning: {num_duplicates} rows with duplicate Links detected.')

if num_duplicates_tc == 0:
    print('No duplicate Job Title and Company combinations detected.')
else:
    print(f'Warning: {num_duplicates_tc} rows share the same Job Title and Company.')
```

```python
# Sample a few records to verify the integrity of the Salary_Raw and JD & Requirements fields

sample_raw = df_raw[['Salary_Raw', 'JD & Requirements']].sample(n=3, random_state=42)

print('RAW DATA STATE VERIFICATION: \n')
for index, row in sample_raw.iterrows():
    print(f'Record Index: {index}')
    salary_val = row['Salary_Raw']
    print(f'Salary_Raw field: {salary_val}')
    jd_val = row['JD & Requirements'][:150]
    jd_snippet = repr(jd_val)
    print(f'JD and Requirements field with hidden characters exposed:')
    print(f'{jd_snippet} [...]\n')
```

## Conclusion
The dataset Data_Raw.csv has been confirmed to be structurally intact, with no duplicate records and no preprocessing applied. The data contains raw text, inconsistent formatting, and hidden characters, which is expected at this stage.
